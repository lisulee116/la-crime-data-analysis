```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

# --- Load dataset ---
df = pd.read_csv("CrimeData.csv")

# --- Basic info ---
print("Shape:", df.shape)
display(df.head(15))
df.info()

# --- Missing values summary ---
print("\nMissing values (top 15):")
print(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
# --- Clean and Standardize Columns ---
df_clean = df.copy()

# Remove redundant columns
df_clean.drop(columns=["Crm Cd 1"], inplace=True, errors="ignore")

# Convert date columns to datetime (explicit format)
DATE_FORMAT = "%m/%d/%Y %I:%M:%S %p"

for col in ["DATE OCC", "Date Rptd"]:
    df_clean[col] = pd.to_datetime(df_clean[col], format=DATE_FORMAT, errors="coerce")

# Ensure LAT/LON are numeric
for col in ["LAT", "LON"]:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

# Check results
df_clean.info()
df_clean[["DATE OCC", "Date Rptd", "TIME OCC", "LAT", "LON"]].head(15)
```

```python
# Pad TIME OCC values to 4 digits (e.g. 1 -> 0001)
df_clean["TIME OCC"] = df_clean["TIME OCC"].astype(str).str.zfill(4)

# Convert to readable "HH:MM" format
df_clean["TIME OCC"] = df_clean["TIME OCC"].str.replace(
    r"(\d{2})(\d{2})", r"\1:\2", regex=True
)

# Quick sanity check
df_clean[["DATE OCC", "TIME OCC"]].head(15)
```

```python
# --- Handle Missing Values and Duplicates ---

# Remove duplicate rows (if any)
dup_count = df_clean.duplicated().sum()
print("Duplicate rows:", dup_count)
if dup_count > 0:
    df_clean = df_clean.drop_duplicates()

# Fill categorical missing values with 'Unknown'
for col in ["AREA NAME", "Crm Cd Desc", "Status Desc", "LOCATION", "Cross Street"]:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna("Unknown")

# Check columns with high missing ratio (>50%)
missing_ratio = df_clean.isna().sum() / len(df_clean) * 100
high_missing = missing_ratio[missing_ratio > 50].sort_values(ascending=False)
print("\nColumns with >50% missing values:")
print(high_missing)

# Quick check again
df_clean.info()
```

```python
# --- Combine Crm Cd 2–4 into a single column ---
crm_cols = ["Crm Cd 2", "Crm Cd 3", "Crm Cd 4"]

# Create 'Additional Crm Cd' by combining non-null values
df_clean["Additional Crm Cd"] = df_clean[crm_cols].apply(
    lambda row: ", ".join([str(int(x)) for x in row if not pd.isna(x)]),
    axis=1
)

# Drop original Crm Cd 2–4 columns
df_clean.drop(columns=crm_cols, inplace=True)

# Check result
df_clean[["Crm Cd", "Additional Crm Cd", "Crm Cd Desc"]].head(10)
```

```python
# --- Consolidate crime codes with primary precedence (robust to missing 'Additional Crm Cd') ---

import numpy as np
import pandas as pd
import re

# 0) If 'Additional Crm Cd' is missing but 2–4가 남아있다면 즉석 생성
crm_cols = ["Crm Cd 2", "Crm Cd 3", "Crm Cd 4"]
if "Additional Crm Cd" not in df_clean.columns and any(c in df_clean.columns for c in crm_cols):
    df_clean["Additional Crm Cd"] = df_clean[crm_cols].apply(
        lambda row: ", ".join([str(int(x)) for x in row if pd.notna(x)]),
        axis=1
    )

# 1) Build code -> description dictionary (from primary rows)
code2desc = (
    df_clean[["Crm Cd", "Crm Cd Desc"]]
    .dropna()
    .drop_duplicates(subset=["Crm Cd"])
    .set_index("Crm Cd")["Crm Cd Desc"]
    .to_dict()
)

# 2) Parser for "860, 420" -> [860, 420]
def parse_additional(s):
    if pd.isna(s) or str(s).strip() == "":
        return []
    return [int(t) for t in re.findall(r"\d{1,4}", str(s))]

# 3) Build add_lists from 'Additional Crm Cd' if present; otherwise empty
add_lists = (
    df_clean["Additional Crm Cd"].apply(parse_additional)
    if "Additional Crm Cd" in df_clean.columns
    else pd.Series([[]] * len(df_clean), index=df_clean.index)
)

# 4) Primary first, then additional (dedup, keep order)
def unify_codes(primary_code, extras):
    out = []
    if pd.notna(primary_code):
        out.append(int(primary_code))
    for c in extras or []:
        if c not in out:
            out.append(c)
    return out

df_clean["All_Crm_Cd_list"] = [unify_codes(pc, ex) for pc, ex in zip(df_clean["Crm Cd"], add_lists)]

# 5) Descriptions for unified list
def codes_to_desc(code_list):
    if not code_list:
        return ""
    seen, descs = set(), []
    for c in code_list:
        d = code2desc.get(c, f"Code {c}")
        if d not in seen:
            seen.add(d)
            descs.append(d)
    return ", ".join(descs)

df_clean["All_Crm_Cd_Desc"] = df_clean["All_Crm_Cd_list"].apply(codes_to_desc)

# 6) Flags + final string column
df_clean["Has_Multiple_Crimes"] = df_clean["All_Crm_Cd_list"].apply(lambda xs: len(xs) > 1)
df_clean["All_Crm_Cd"] = df_clean["All_Crm_Cd_list"].apply(lambda xs: ", ".join(str(x) for x in xs))

# 7) Preview BEFORE dropping helpers (to verify multiples show up)
print(df_clean[["Crm Cd", "Additional Crm Cd"]].head(10) if "Additional Crm Cd" in df_clean.columns else "No 'Additional Crm Cd' present")
print(df_clean[["Crm Cd", "All_Crm_Cd", "All_Crm_Cd_Desc"]].head(10))
print("Share of multi-offense rows:", float(df_clean["Has_Multiple_Crimes"].mean()))

# 8) Drop helpers
df_clean.drop(columns=["All_Crm_Cd_list", "Additional Crm Cd", *[c for c in crm_cols if c in df_clean.columns]], inplace=True, errors="ignore")
df_clean[["Crm Cd", "All_Crm_Cd", "All_Crm_Cd_Desc"]].head(10)
```

```python
# --- Feature Engineering: Time and Date ---

# Derive year, month, day of week, weekend flag
df_clean["YEAR"] = df_clean["DATE OCC"].dt.year
df_clean["MONTH"] = df_clean["DATE OCC"].dt.month
df_clean["DAY_OF_WEEK"] = df_clean["DATE OCC"].dt.day_name()
df_clean["IS_WEEKEND"] = df_clean["DAY_OF_WEEK"].isin(["Saturday", "Sunday"])

# Derive time period from TIME OCC (without changing it)
def to_time_period(v):
    try:
        hour = int(float(v)) // 100
    except:
        return "Unknown"
    if 5 <= hour < 12: return "Morning"
    if 12 <= hour < 18: return "Afternoon"
    if 18 <= hour < 24: return "Evening"
    if 0 <= hour < 5: return "Night"
    return "Unknown"

df_clean["TIME_PERIOD"] = df_clean["TIME OCC"].apply(to_time_period)

# Check result
df_clean[["DATE OCC", "TIME OCC", "TIME_PERIOD", "DAY_OF_WEEK", "IS_WEEKEND"]].head(10)
```

```python
import pandas as pd, re

MAP_PATH = "MoCodes.csv"  # your uploaded file

# Load and normalize columns
mo_map_df = pd.read_csv(MAP_PATH, dtype=str, encoding_errors="ignore")
mo_map_df.columns = (
    mo_map_df.columns
    .str.replace("\ufeff", "", regex=False)  # strip BOM
    .str.strip()
    .str.lower()
)

print("Columns in MoCodes.csv:", list(mo_map_df.columns))
display(mo_map_df.head(10))
print("Rows in mapping csv:", len(mo_map_df))

# Try to find code/description columns by common names
code_col_candidates = [c for c in mo_map_df.columns if c in ["code","codes","mocode","mo code","mo_code","mo codes","mo_codes"]]
desc_col_candidates = [c for c in mo_map_df.columns if c in ["description","desc","details","text"]]

print("Detected code cols:", code_col_candidates)
print("Detected desc cols:", desc_col_candidates)

# Pick the first plausible columns (adjust if needed)
CODE_COL = code_col_candidates[0] if code_col_candidates else "code"
DESC_COL = desc_col_candidates[0] if desc_col_candidates else "description"

# If columns are named differently, try to rename smartly
if CODE_COL not in mo_map_df.columns or DESC_COL not in mo_map_df.columns:
    # Heuristic: if exactly two columns, assume first is code, second is description
    if len(mo_map_df.columns) == 2:
        mo_map_df = mo_map_df.rename(columns={mo_map_df.columns[0]:"code", mo_map_df.columns[1]:"description"})
        CODE_COL, DESC_COL = "code", "description"
    else:
        print("⚠️ Couldn't auto-detect columns. Please confirm column names.")

# Clean and normalize code keys to 4-digit strings
def normalize_code(s):
    if s is None: return None
    s = str(s)
    # keep digits only (handles '0377,' '0377.' etc.)
    m = re.search(r"(\d{1,4})", s)
    if not m: return None
    return m.group(1).zfill(4)

mo_map_df = mo_map_df.dropna(subset=[CODE_COL, DESC_COL])
mo_map_df["__code"] = mo_map_df[CODE_COL].apply(normalize_code)
mo_map_df["__desc"] = mo_map_df[DESC_COL].astype(str).str.strip()

# Drop rows that failed to normalize
mo_map_df = mo_map_df.dropna(subset=["__code","__desc"]).drop_duplicates("__code", keep="first")

# Build dict
mo_map_str = dict(zip(mo_map_df["__code"], mo_map_df["__desc"]))

print("Mapping size:", len(mo_map_str))
print("Has 0377?", "0377" in mo_map_str)
if "0377" in mo_map_str:
    print("0377 ->", mo_map_str["0377"])

# Peek a few keys
print("Some keys:", list(mo_map_str.keys())[:15])
```

```python
# Strict decoder: extract only digits, pad to 4, then lookup
def decode_mocodes_with_csv(val, mapping):
    if pd.isna(val):
        return []
    # replace any non-digit with space, then split
    tokens = re.sub(r"\D+", " ", str(val)).strip().split()
    out = []
    for t in tokens:
        key = t.zfill(4)
        out.append(mapping.get(key, f"Unknown({key})"))
    return out

df_clean["MoCodes_desc"] = df_clean["Mocodes"].apply(
    lambda v: ", ".join(decode_mocodes_with_csv(v, mo_map_str))
)

df_clean[["Mocodes", "MoCodes_desc"]].head(10)
```

```python
# Drop the decoded list column
df_clean = df_clean.drop(columns=["MoCodes_decoded"], errors="ignore")

# Check results
df_clean[["Mocodes", "MoCodes_desc"]].head()
```

```python
print("Vict Sex unique values:", sorted(df_clean["Vict Sex"].dropna().unique()))
print("Vict Descent unique values:", sorted(df_clean["Vict Descent"].dropna().unique()))
print("\nMissing counts:")
print(df_clean[["Vict Sex","Vict Descent"]].isna().sum())
```

```python
# --- Normalize Vict Sex ---
valid_sex = {"M", "F"}

# keep NaN as NaN (use pandas' string dtype), normalize case/whitespace
df_clean["Vict Sex"] = (
    df_clean["Vict Sex"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# map explicit invalid codes to UNKNOWN
df_clean["Vict Sex"] = df_clean["Vict Sex"].replace({"X": "UNKNOWN", "-": "UNKNOWN"})

# allow only M/F/UNKNOWN; everything else -> UNKNOWN, then fill NaN as UNKNOWN
df_clean["Vict Sex"] = (
    df_clean["Vict Sex"]
    .where(df_clean["Vict Sex"].isin(valid_sex | {"UNKNOWN"}), "UNKNOWN")
    .fillna("UNKNOWN")
)

# final presentation casing
df_clean["Vict Sex"] = df_clean["Vict Sex"].replace({"UNKNOWN": "Unknown"})

# --- Normalize Vict Descent ---
# LAPD descent codes observed (exclude '-', 'X')
valid_descent = {"A","B","C","D","F","G","H","I","J","K","L","O","P","S","U","V","W","Z"}

df_clean["Vict Descent"] = (
    df_clean["Vict Descent"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df_clean["Vict Descent"] = df_clean["Vict Descent"].replace({"X": "UNKNOWN", "-": "UNKNOWN"})

df_clean["Vict Descent"] = (
    df_clean["Vict Descent"]
    .where(df_clean["Vict Descent"].isin(valid_descent | {"UNKNOWN"}), "UNKNOWN")
    .fillna("UNKNOWN")
)
df_clean["Vict Descent"] = df_clean["Vict Descent"].replace({"UNKNOWN": "Unknown"})

# --- Quick checks ---
print("Vict Sex ->", df_clean["Vict Sex"].value_counts(dropna=False))
print("Vict Descent ->", df_clean["Vict Descent"].value_counts(dropna=False).head(15))

df_clean[["Vict Age", "Vict Sex", "Vict Descent"]].head(15)
```

```python
# Mapping from LAPD official metadata
descent_map = {
    "A": "Other Asian",
    "B": "Black",
    "C": "Chinese",
    "D": "Cambodian",
    "F": "Filipino",
    "G": "Guamanian",
    "H": "Hispanic/Latin/Mexican",
    "I": "American Indian/Alaskan Native",
    "J": "Japanese",
    "K": "Korean",
    "L": "Laotian",
    "O": "Other",
    "P": "Pacific Islander",
    "S": "Samoan",
    "U": "Hawaiian",
    "V": "Vietnamese",
    "W": "White",
    "X": "Unknown",
    "Z": "Asian Indian",
    "-": "Unknown",
    "Unknown": "Unknown"
}

# Apply mapping — creates a new descriptive column
df_clean["Vict Descent Full"] = df_clean["Vict Descent"].map(descent_map).fillna("Unknown")

# keep both (for reference)
df_clean[["Vict Age", "Vict Sex", "Vict Descent", "Vict Descent Full"]].head(15)
```
